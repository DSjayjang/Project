```python
x_cp = [0.1, 0.5, 1.0, 1.5, 2.0]
y_cp_acc = [93.12, 92.42, 93.38, 92.75, 92.48]

x_sparse = [0.1, 0.3, 0.5, 0.8, 1.0]
y_sparse_acc = [92.52, 92.60, 93.38, 92.95, 92.30]

x_diverse = [0.01, 0.05, 0.1, 0.5, 1.0]
y_diverse_acc = [93.32, 92.58, 93.38, 92.35, 93.18]

xticks = [0.01, 0.1, 0.3, 0.5, 0.8, 1.0, 1.5, 2.0]
yticks = [92.0, 92.5, 93.0, 93.38, 94.0]


fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey = True)
# -------------------------------------------------------------------------
# Subplot (a): Class Prototype Margin (lambda_cp)
# -------------------------------------------------------------------------
axes[0].plot(x_cp, y_cp_acc, marker='o', linestyle='-', linewidth=2.2, color='seagreen', label=r'$\lambda_{CP}$ ($\lambda_{S}=0.5, \lambda_{D}=0.1$)')
axes[0].set_xticks(x_cp)
axes[0].set_xlabel('$\lambda_{CP}$', fontsize=15)
axes[0].set_ylabel('Top-1 Accuracy (%)', fontsize=15)
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend(frameon=True, facecolor='white', edgecolor='gainsboro', fontsize=13, loc='upper right')
for x, y in zip(x_cp, y_cp_acc):
    axes[0].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0, 7), ha='center', va='bottom', fontsize=9)

# -------------------------------------------------------------------------
# Subplot (b): Multi-head Sparsity Regularization (lambda_sparse)
# -------------------------------------------------------------------------
axes[1].plot(x_sparse, y_sparse_acc, marker='s', linestyle='-', linewidth=2.2, color='steelblue', label=r'$\lambda_{S}$ ($\lambda_{CP}=1.0, \lambda_{D}=0.1$)')
axes[1].set_xticks(x_sparse)
axes[1].set_xlabel('$\lambda_{S}$', fontsize=15)
axes[1].grid(True, linestyle='--', alpha=0.5)
axes[1].legend(frameon=True, facecolor='white', edgecolor='gainsboro', fontsize=13, loc='upper right')
for x, y in zip(x_sparse, y_sparse_acc):
    axes[1].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0, 7), ha='center', va='bottom', fontsize=9)

# -------------------------------------------------------------------------
# Subplot (c): Multi-head Diversity Regularization (lambda_diverse)
# -------------------------------------------------------------------------
axes[2].plot(x_diverse, y_diverse_acc, marker='^', linestyle='-', linewidth=2.2, color='darkorange', label=r'$\lambda_{D}$ ($\lambda_{CP}=1.0, \lambda_{S}=0.5$)')
axes[2].set_xticks(x_diverse)
axes[2].set_xlabel('$\lambda_{D}$', fontsize=15,)
axes[2].grid(True, linestyle='--', alpha=0.5)
axes[2].legend(frameon=True, facecolor='white', edgecolor='gainsboro', fontsize=13, loc='upper right')
for x, y in zip(x_diverse, y_diverse_acc):
    axes[2].annotate(f'{y}', (x, y), textcoords="offset points", xytext=(0, 7), ha='center', va='bottom', fontsize=9)

plt.ylim(90.0, 95.0)
# fig.suptitle('Hyperparameter Sensitivity', fontsize=14, fontweight='bold', y=1.02)
fig.text(0.20, -0.05, "(a) Class-prototype prior weight $\lambda_{CP}$", ha="center", fontsize=18)
fig.text(0.52, -0.05, "(b) Sparsity regularizer weight $\lambda_{S}$", ha="center", fontsize=18)
fig.text(0.83, -0.05, "(c) Diversity regularizer weight $\lambda_{D}$", ha="center", fontsize=18)

plt.tight_layout()
plt.savefig('hyperparameter_sensitivity_subplots.png', bbox_inches='tight', dpi=300)
plt.show()
```
